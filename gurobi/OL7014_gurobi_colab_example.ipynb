```python
# 2) 수업용 라이선스로 Gurobi 환경 만들기
import gurobipy as gp
from gurobipy import GRB

params = {
    "WLSACCESSID": "여기에 ACCESSID 붙여넣기",
    "WLSSECRET":   "여기에 SECRET 붙여넣기",
    "LICENSEID":   123456,   # 숫자 그대로, 따옴표 없이
}
env = gp.Env(params=params)
```

```python
# 3) 예제: 0-1 배낭 문제 (knapsack)
value  = [10, 13, 18, 31, 7, 15]   # 물건의 가치
weight = [ 2,  3,  4,  5, 1,  3]   # 물건의 무게
W = 10                             # 배낭 용량
n = len(value)

m = gp.Model("knapsack", env=env)
x = m.addVars(n, vtype=GRB.BINARY, name="x")
m.setObjective(gp.quicksum(value[i] * x[i] for i in range(n)), GRB.MAXIMIZE)
m.addConstr(gp.quicksum(weight[i] * x[i] for i in range(n)) <= W)
m.optimize()

print(f"\n최적값: {m.ObjVal:g}")   # 56이 나오면 성공
print("선택한 물건:", [i for i in range(n) if x[i].X > 0.5])
```
